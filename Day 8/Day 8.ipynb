The New Jerusalem - Revelations 21:1 - 22:5

```python
"""
The New Jerusalem — Revelation 21:1–22:5, drawn to scale in km
#30DayMapChallenge · Utopia

Renders two versions:
  new_jerusalem_verses.png  — labels quote the text, with verse references
  new_jerusalem_clean.png   — short labels only, no verses

Requires: matplotlib, numpy
"""
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle, FancyBboxPatch, Polygon
from matplotlib.colors import LinearSegmentedColormap

# ---- Measurements ------------------------------------------------------------
FURLONG_KM = 0.185                     # 1 stadion (NIV "stadia", KJV "furlong")
SIDE = 12_000 * FURLONG_KM            # Rev 21:16 → ~2,220 km
HALF = SIDE / 2
BAND = 28                              # drawn width of each foundation band (km)

FOUNDATIONS = [                        # Rev 21:19–20 (NIV names), first = outermost
    ("Jasper", "#9e2b25"), ("Sapphire", "#1f4e9c"), ("Agate", "#9fb3c8"),
    ("Emerald", "#1f8a5b"), ("Onyx", "#4a4246"), ("Ruby", "#b3261e"),
    ("Chrysolite", "#c8b23a"), ("Beryl", "#5fb3a6"), ("Topaz", "#e3a72f"),
    ("Turquoise", "#2fb5c9"), ("Jacinth", "#d9692b"), ("Amethyst", "#7b4aa0"),
]
GATES = {                              # Ezekiel 48:31–34
    "N": ["Reuben", "Judah", "Levi"],
    "E": ["Joseph", "Benjamin", "Dan"],
    "S": ["Simeon", "Issachar", "Zebulun"],
    "W": ["Gad", "Asher", "Naphtali"],
}

PARCHMENT = "#f3ead7"
EARTH = "#e4e3c4"
INK = "#3b2f22"
GOLD = "#e8c766"
RIVER = "#bfe6f2"
SERIF = dict(family="serif", color=INK)
glory = LinearSegmentedColormap.from_list("glory", ["#d9b24c", "#f2dc8f", "#fffbe8"])
LABEL_BOX = dict(boxstyle="round,pad=0.25", fc="#fffaf0", ec="none", alpha=0.85)

# ---- Label text for each version ---------------------------------------------
TEXT = {
    True: dict(
        earth="THE NEW EARTH  ·  “and there was no more sea” (21:1)",
        mountain="a great and high mountain:\nwhence John saw the city descend (21:2, 10)",
        throne="THE THRONE OF GOD AND OF THE LAMB\nin the midst of the city (22:1, 3)",
        river_tree="RIVER OF THE WATER OF LIFE, clear as crystal,\nout of the throne (22:1)\n\n"
                   "THE TREE OF LIFE, on either side of the river;\ntwelve fruits, one each month;\n"
                   "leaves for the healing of the nations (22:2)",
        notes=[(-640, 640, "NO TEMPLE: God and the Lamb are its temple (21:22)\n"
                           "NO NIGHT: no sun or moon, the Lamb is its light (21:23, 25)"),
               (-640, -560, "STREET OF PURE GOLD,\nas it were transparent glass (21:21)\n\n"
                            "His servants shall see his face;\nhis name on their foreheads (22:4)")],
        gates="Pearl gate · twelve in all, each named for a tribe of Israel (21:12–13, 21)",
        legend="TWELVE FOUNDATIONS · names of the twelve apostles (21:14, 19–20)",
        subtitle="“And I John saw the holy city, new Jerusalem, coming down from God out of heaven” (Rev 21:2)",
        dims="12,000 furlongs square ≈ 2,220 km  ·  wall of 144 cubits ≈ 66 m (21:16–17)",
        foot="Gate names after Ezekiel 48:31–34; river flowing east after Ezekiel 47:1 (Revelation fixes neither). "
             "The tree of life is drawn as one tree\non either side of the river (22:2). Gem colours approximate. "
             "Quotations KJV.   #30DayMapChallenge · Utopia",
    ),
    False: dict(
        earth="THE NEW EARTH",
        mountain="the high mountain",
        throne="",
        river_tree="",
        notes=[],
        gates="Pearl gate",
        legend="TWELVE FOUNDATIONS (Rev 21:19–20)",
        subtitle="The holy city of Revelation 21–22, drawn to scale",
        dims="12,000 stadia square ≈ 2,220 km  ·  wall 144 cubits thick ≈ 66 m",
        foot="Gate names after Ezekiel 48. Tree of life drawn as one tree spanning the river. "
             "Gem colours approximate. Scripture quotations NIV.   #30DayMapChallenge · Utopia",
    ),
}

offs = [-SIDE / 4, 0, SIDE / 4]


def gate_xy(s, k):
    o = offs[k]
    return {"N": (o, HALF), "S": (o, -HALF), "E": (HALF, -o), "W": (-HALF, -o)}[s]


def draw(verses, outfile):
    T = TEXT[verses]
    rng = np.random.default_rng(12)            # same random detail in both versions

    fig = plt.figure(figsize=(13.5, 15), dpi=200)
    fig.patch.set_facecolor(PARCHMENT)
    ax = fig.add_axes([0.01, 0.045, 0.98, 0.86])
    ax.set_facecolor(PARCHMENT)
    ax.set_aspect("equal")
    pad = 430
    ax.set_xlim(-HALF - pad, HALF + pad)
    ax.set_ylim(-HALF - pad - 200, HALF + pad - 60)
    ax.axis("off")

    # The new earth around the city (21:1)
    E = HALF + 330
    ax.add_patch(FancyBboxPatch((-E, -E), 2 * E, 2 * E, boxstyle="round,pad=0,rounding_size=160",
                                fc=EARTH, ec="#c9c7a2", lw=1, zorder=0))
    for _ in range(900):                                  # meadow stipple
        x, y = rng.uniform(-E + 40, E - 40, 2)
        if max(abs(x), abs(y)) > HALF + 60:
            ax.plot(x, y, ".", ms=1.2, color="#a9ad7a", alpha=.6, zorder=0)

    # Foundations
    for i, (_, col) in enumerate(FOUNDATIONS):
        h = HALF - i * BAND
        ax.add_patch(Rectangle((-h, -h), 2 * h, 2 * h, fc=col, ec="none", zorder=4))
    inner = HALF - len(FOUNDATIONS) * BAND

    # Gold floor lit from the throne (21:23)
    xs = np.linspace(-inner, inner, 500)
    X, Y = np.meshgrid(xs, xs)
    ax.imshow(np.exp(-(X**2 + Y**2) / (2 * (inner * .55) ** 2)), extent=(-inner, inner, -inner, inner),
              cmap=glory, origin="lower", zorder=5, interpolation="bicubic")
    clip = Rectangle((-inner, -inner), 2 * inner, 2 * inner, transform=ax.transData)
    for a in np.linspace(0, 2 * np.pi, 48, endpoint=False):
        ln, = ax.plot([0, inner * 1.4 * np.cos(a)], [0, inner * 1.4 * np.sin(a)],
                      color="white", lw=.6, alpha=.35, zorder=5)
        ln.set_clip_path(clip)

    # Streets of gold from each gate to the throne (21:21)
    for s in GATES:
        for k in range(3):
            gx, gy = gate_xy(s, k)
            ax.plot([gx, 0], [gy, 0], color=GOLD, lw=2.2, alpha=.7, zorder=6)
            ax.plot([gx, 0], [gy, 0], color="#fff6d0", lw=.6, alpha=.9, zorder=6)

    # River of the water of life, eastward from the throne (22:1; cf. Ezek 47:1)
    t = np.linspace(0, 1, 400)
    rx = t * (HALF - 10)
    ry = 18 * np.sin(t * 9 * np.pi) * t
    ax.plot(rx, ry, color="#7fb9cf", lw=9, solid_capstyle="round", zorder=7)
    ax.plot(rx, ry, color=RIVER, lw=6, solid_capstyle="round", zorder=7)

    # The tree of life, one tree on either side of the river (22:2)
    TX, TR = 480, 112
    tyc = 18 * np.sin(TX / (HALF - 10) * 9 * np.pi) * TX / (HALF - 10)
    for sgn in (-1, 1):                                    # roots on both banks
        for k in range(7):
            a0 = np.radians(sgn * 90 + rng.uniform(-55, 55))
            r = np.linspace(20, TR + rng.uniform(25, 60), 30)
            wob = np.cumsum(rng.normal(0, .018, 30))
            ax.plot(TX + r * np.cos(a0 + wob), tyc + r * np.sin(a0 + wob),
                    color="#6b4a2b", lw=rng.uniform(1.2, 2.6), solid_capstyle="round", zorder=8)
    ax.add_patch(Circle((TX, tyc), TR, fc="#2f6b3a", ec="#1f4a28", lw=1.2, alpha=.93, zorder=9))
    for _ in range(40):                                    # leaf clusters
        rr, aa = TR * np.sqrt(rng.uniform(0, .8)), rng.uniform(0, 2 * np.pi)
        ax.add_patch(Circle((TX + rr * np.cos(aa), tyc + rr * np.sin(aa)), rng.uniform(14, 26),
                            fc=rng.choice(["#3f8a48", "#4f9a52", "#2a5f34"]), ec="none", alpha=.8, zorder=9))
    for m in range(12):                                    # twelve fruits, one each month
        a = np.pi / 2 - 2 * np.pi * m / 12
        ax.add_patch(Circle((TX + .72 * TR * np.cos(a), tyc + .72 * TR * np.sin(a)), 11,
                            fc=FOUNDATIONS[m][1], ec="#fffaf0", lw=1, zorder=10))
    ax.add_patch(Circle((TX, tyc), 16, fc="#6b4a2b", ec="#4a321c", lw=.8, zorder=10))

    # Throne of God and of the Lamb (22:1, 3)
    ax.add_patch(Circle((0, 0), 46, fc="white", ec=GOLD, lw=3, zorder=10))
    ax.add_patch(Circle((0, 0), 64, fc="none", ec="#1f8a5b", lw=2, zorder=10))

    # Labels inside the city
    if T["throne"]: ax.text(80, 98, T["throne"], fontsize=7.5 if verses else 9, weight="bold", va="bottom", ha="right",
            zorder=12, bbox=LABEL_BOX, **SERIF)
    if T["river_tree"]: ax.text(TX + 40, -TR - 75, T["river_tree"], fontsize=7.5 if verses else 9, ha="center",
            va="top", zorder=12, bbox=LABEL_BOX, **SERIF)
    for x, y, txt in T["notes"]:
        ax.text(x, y, txt, fontsize=7.5, zorder=12, bbox=LABEL_BOX, **SERIF)

    ax.add_patch(Rectangle((-HALF, -HALF), SIDE, SIDE, fc="none", ec=INK, lw=.8, zorder=12))

    # Pearl gates
    for s, tribes in GATES.items():
        for k, tribe in enumerate(tribes):
            gx, gy = gate_xy(s, k)
            ax.add_patch(Circle((gx, gy), 42, fc="#fbf8f2", ec="#cfc6b8", lw=1.5, zorder=13))
            ax.add_patch(Circle((gx - 12, gy + 12), 12, fc="white", alpha=.9, zorder=14))
            dx, dy, rot = {"N": (0, 64, 0), "S": (0, -64, 0), "E": (66, 0, -90), "W": (-66, 0, 90)}[s]
            ax.text(gx + dx, gy + dy, tribe.upper(), rotation=rot, ha="center",
                    va={"N": "bottom", "S": "top"}.get(s, "center"), fontsize=13.5, zorder=14,
                    bbox=dict(boxstyle="round,pad=0.15", fc=EARTH, ec="none"), **SERIF)

    # Scale bar
    sb_y = -E - 110
    for i in range(4):
        ax.add_patch(Rectangle((-HALF + i * 250, sb_y), 250, 30, fc=INK if i % 2 == 0 else PARCHMENT,
                               ec=INK, lw=.8))
    for i in range(5):
        ax.text(-HALF + i * 250, sb_y - 25, f"{i*250}", ha="center", va="top", fontsize=12, **SERIF)
    ax.text(-HALF + 1035, sb_y + 15, "km", va="center", fontsize=12, **SERIF)

    # Gate legend, under the scale bar
    gl_x, gl_y = -HALF + 30, sb_y - 140
    ax.add_patch(Circle((gl_x, gl_y), 30, fc="#fbf8f2", ec="#cfc6b8", lw=1.4))
    ax.add_patch(Circle((gl_x - 8, gl_y + 8), 8, fc="white", alpha=.9))
    ax.text(gl_x + 55, gl_y, T["gates"], va="center", fontsize=11, **SERIF)

    # Foundations legend
    lx, ly = -HALF + 1200, sb_y + 25
    ax.text(lx, ly, T["legend"], fontsize=11, weight="bold", **SERIF)
    quote = ""
    top = ly - 70
    if quote:
        ax.text(lx, ly - 30, quote, fontsize=10.5, style="italic", va="top", **SERIF)
        top = ly - 175
    for i, (name, col) in enumerate(FOUNDATIONS):
        r, c = divmod(i, 3)
        x, y = lx + c * 400, top - r * 58
        ax.add_patch(Rectangle((x, y), 38, 38, fc=col, ec=INK, lw=.5, clip_on=False))
        ax.text(x + 52, y + 19, f"{i+1}. {name}", va="center", fontsize=11, **SERIF)

    # Titles + footnote
    fig.text(0.5, 0.968, "THE NEW JERUSALEM", ha="center", fontsize=32, weight="bold", **SERIF)
    fig.text(0.5, 0.945, T["subtitle"], ha="center", fontsize=11.5, style="italic", **SERIF)
    fig.text(0.5, 0.926, T["dims"], ha="center", fontsize=10.5, **SERIF)
    fig.text(0.02, 0.008, T["foot"], fontsize=11, style="italic", **SERIF)

    plt.savefig(outfile, dpi=200, facecolor=PARCHMENT)
    plt.close(fig)
    print("saved", outfile)


if __name__ == "__main__":
    draw(verses=True, outfile="new_jerusalem_verses.png")
    draw(verses=False, outfile="new_jerusalem_clean.png")
```
