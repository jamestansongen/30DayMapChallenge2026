# Day 1: UK Trigpoints (2018)

Data Source: https://github.com/philmikejones/trigpoints

```python
import warnings
warnings.filterwarnings("ignore", category=UserWarning)   # hide the rdata warnings

import rdata
import geopandas as gpd

df = rdata.read_rda("trigpoints.rda")["trigpoints"]

# Split the [easting, northing] pairs into two columns
df["easting"] = [g[0] for g in df["geometry"]]
df["northing"] = [g[1] for g in df["geometry"]]

# Keep surviving pillars only
pillars = df[(df["type"] == "PILLAR") & (df["destroyed"] == 0)].copy()
pillars = pillars[["name", "station", "height", "order", "maintained", "easting", "northing"]]

# Make proper points in British National Grid and save
gdf = gpd.GeoDataFrame(
    pillars,
    geometry=gpd.points_from_xy(pillars["easting"], pillars["northing"]),
    crs=27700,
)
gdf.to_file("trig_pillars_gb.geojson", driver="GeoJSON")
print(len(gdf), "pillars saved")
```
